```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.metrics import mean_squared_error,r2_score
from sklearn.preprocessing import StandardScaler
```

```python
diabetes = load_diabetes()
x = diabetes.data
y = diabetes.target
```

```python
x_train,x_test,y_train ,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
print("Training samples:",x_train.shape[0])
print("Test samples:",x_test.shape[0])
```

```python
linear_model = LinearRegression()
linear_model.fit(x_train_scaled,y_train)
y_pred_linear = linear_model.predict(x_test_scaled)
print("Linear Regression:")
print("Mean Squared Error:",mean_squared_error(y_test,y_pred_linear))
print("R2 Score:",r2_score(y_test,y_pred_linear))
```

```python
ridge = Ridge()
param_grid = {'alpha':[0.01,0.1,1,10,100]}
ridge_cv = GridSearchCV(ridge,param_grid,cv=5)
ridge_cv.fit(x_train_scaled,y_train)
best_ridge = ridge_cv.best_estimator_
y_pred_ridge = best_ridge.predict(x_test_scaled)
```

```python
print("Ridge Regression:")
print("Best Alpha:",ridge_cv.best_params_['alpha'])
print("Mean Squared Error:",mean_squared_error(y_test,y_pred_ridge))
print("R2 Score:",r2_score(y_test,y_pred_ridge))
```

```python
lasso = Lasso(max_iter=10000)
lasso_cv =GridSearchCV(lasso,param_grid,cv=5)
lasso_cv.fit(x_train_scaled,y_train)

best_lasso = lasso_cv.best_estimator_
y_pred_lasso = best_lasso.predict(x_test_scaled)

print("Best Alpa for Lasso:",lasso_cv.best_params_['alpha'])
print("Mean Squared Error:",mean_squared_error(y_test,y_pred_lasso))
print("R2 Score:",r2_score(y_test,y_pred_lasso))
```

```python
results = pd.DataFrame({
    'Model':['Linear Regression','Ridge Regression','Lasso Regression'],
    'Mean Squared Error':[mean_squared_error(y_test,y_pred_linear),mean_squared_error(y_test,y_pred_ridge),mean_squared_error(y_test,y_pred_lasso)],"R2 Score":[r2_score(y_test,y_pred_linear),r2_score(y_test,y_pred_ridge),r2_score(y_test,y_pred_lasso)]
})
results
```

```python
plt.figure(figsize=(8,5))
models = ['Linear Regression','Ridge Regression','Lasso Regression']
r2_scores = [r2_score(y_test,y_pred_linear),r2_score(y_test,y_pred_ridge),r2_score(y_test,y_pred_lasso)]
plt.bar(models,r2_scores)
plt.xlabel('Models')
plt.ylabel('R2 Score')
plt.title('Model Comparison')
plt.show()
```
